```python
import cv2
import matplotlib.pyplot as plt
```

grayscale

flags for imread:

- cv2.IMREAD_COLOR or 1 – default
- cv2.IMREAD_GRAYSCALE or 0
- cv2.IMREAD_UNCHANGED or -1

```python
image_gray = cv2.imread('images/FCK.png', 0)
```

```python
# need to tell matplotlib to plot grayscale
plt.imshow(image_gray, cmap='gray', vmin=0, vmax=255)
```

RGB color channels

```python
image = cv2.imread('images/FCK.png')

image[450:600, 450:600]
```

OpenCV represents the images in BGR as opposed to the RGB we expect

```python
image[450:600, 450:600, ::-1]
```

fix matplotlib color plotting

```python
image_RGB = image[..., ::-1] # reverse BGR in channel dimension, keep spatial dimensions as is
plt.imshow(image_RGB)

# plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
```
